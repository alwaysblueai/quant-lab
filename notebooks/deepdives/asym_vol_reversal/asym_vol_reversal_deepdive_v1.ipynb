# Asym Vol Reversal - Deepdive v1

Tier-2 notebook skeleton for factor-specific checks.
Use only when a factor passes Tier-1 and needs deeper mechanism diagnosis.


```python
from __future__ import annotations

from pathlib import Path
from typing import Iterable

import matplotlib.pyplot as plt
import pandas as pd

from alpha_lab.grouped_evaluation import conditional_ic_by_bucket
from alpha_lab.regime import classify_market_regimes


def _read_table(paths: Iterable[Path], *, required: bool = True) -> pd.DataFrame:
    """Load the first existing artifact path as parquet/csv."""
    for path in paths:
        path = Path(path)
        if not path.exists():
            continue

        if path.suffix.lower() == '.parquet':
            return pd.read_parquet(path)

        if path.suffix.lower() in {'.csv', '.txt'}:
            return pd.read_csv(path)

    if required:
        candidates = [str(path) for path in paths]
        raise FileNotFoundError(f'No matching file for candidates: {candidates}')

    return pd.DataFrame()


def _quantile_labels(series: pd.Series, *, labels: list[str]) -> pd.Series:
    out = pd.Series(pd.NA, index=series.index, dtype='object')
    valid = pd.to_numeric(series, errors='coerce').dropna()
    if len(valid) < len(labels):
        return out
    out.loc[valid.index] = pd.qcut(
        valid.rank(method='first'),
        q=len(labels),
        labels=labels,
    ).astype(str)
    return out


def build_quantile_buckets(
    frame: pd.DataFrame,
    *,
    value_col: str,
    bucket_col: str = 'bucket',
    n_buckets: int = 5,
) -> pd.DataFrame:
    """Build per-date quantile buckets by date and asset."""
    if n_buckets < 2:
        raise ValueError('n_buckets must be >= 2')
    if not isinstance(frame, pd.DataFrame):
        raise TypeError(f'frame must be a DataFrame, got {type(frame).__name__}')

    frame = frame.loc[:, ~frame.columns.duplicated()].copy()
    required = {'date', 'asset', value_col}
    if not required.issubset(frame.columns):
        raise ValueError(f'Expected columns {sorted(required)}')

    out = frame.loc[:, ['date', 'asset']].copy()
    out[value_col] = pd.to_numeric(frame[value_col], errors='coerce')
    out['date'] = pd.to_datetime(out['date'], errors='coerce')
    out = out.dropna(subset=['date', 'asset', value_col]).copy()

    labels = [f'Q{i + 1}' for i in range(int(n_buckets))]
    out[bucket_col] = out.groupby('date', sort=False)[value_col].transform(
        lambda series: _quantile_labels(series, labels=labels)
    )

    return out[['date', 'asset', bucket_col]].dropna(subset=['date', 'asset', bucket_col])


def plot_bucket_series(df: pd.DataFrame, *, title: str) -> None:
    if df.empty:
        print(f'Skip plot: {title} (empty)')
        return

    pivot = df.pivot(index='date', columns='bucket', values='ic').sort_index()
    if pivot.empty:
        print(f'Skip plot: {title} (no date/bucket)')
        return

    pivot.plot(figsize=(11, 4))
    plt.title(title)
    plt.axhline(0.0, color='black', linewidth=0.8, alpha=0.35)
    plt.tight_layout()
    plt.show()
```

## 1. Configure case path and load artifacts


```python
# Set your completed case directory here.
# NOTE: Notebook working directory may not be repo root, so we locate the repo root first.

def _find_repo_root(start: Path) -> Path:
    cur = start.resolve()
    for parent in [cur, *cur.parents]:
        if (parent / 'pyproject.toml').exists():
            return parent
    return cur

REPO_ROOT = _find_repo_root(Path.cwd())

case_name = 'asym_vol_reversal_v1'
case_candidates = [
    REPO_ROOT / 'outputs' / 'single_factor' / case_name,
    REPO_ROOT / 'outputs' / f'case_{case_name}',  # legacy-ish
    REPO_ROOT / 'output' / f'case_{case_name}',   # legacy-ish
    REPO_ROOT / 'output' / case_name,             # legacy-ish
]

CASE_DIR = next((path for path in case_candidates if path.exists()), None)
if CASE_DIR is None:
    existing = []
    base = REPO_ROOT / 'outputs' / 'single_factor'
    if base.exists():
        existing = sorted([p.name for p in base.iterdir() if p.is_dir()])
    raise FileNotFoundError(
        f'Repo root inferred as: {REPO_ROOT}\n\n'
        + 'Case dir not found. Tried candidates:\n'
        + '\n'.join(f'- {p}' for p in case_candidates)
        + (('\n\nExisting under outputs/single_factor:\n' + '\n'.join(f'- {n}' for n in existing)) if existing else '')
    )

print(f'Repo root: {REPO_ROOT}')
print(f'Using CASE_DIR={CASE_DIR}')

factor_df = _read_table([
    CASE_DIR / 'factor_values.parquet',
    CASE_DIR / 'factor_values.csv',
    CASE_DIR / 'factor.parquet',
    CASE_DIR / 'factor.csv',
], required=True)
labels_df = _read_table([
    CASE_DIR / 'labels.parquet',
    CASE_DIR / 'labels.csv',
], required=True)
prices = _read_table([
    CASE_DIR / 'prices.parquet',
    CASE_DIR / 'prices.csv',
    CASE_DIR / 'panel.parquet',
    CASE_DIR / 'panel.csv',
], required=False)
ic_df = _read_table([
    CASE_DIR / 'ic_timeseries.parquet',
    CASE_DIR / 'ic_timeseries.csv',
], required=False)
long_short_df = _read_table([
    CASE_DIR / 'long_short_return.parquet',
    CASE_DIR / 'long_short_return.csv',
], required=False)

for name, frame in {'factor_df': factor_df, 'labels_df': labels_df, 'prices': prices, 'ic_df': ic_df, 'long_short_df': long_short_df}.items():
    print(f'{name}: {frame.shape[0]} x {frame.shape[1]}')

for frame in [factor_df, labels_df]:
    frame['date'] = pd.to_datetime(frame['date'], errors='coerce')

for frame in [prices, ic_df, long_short_df]:
    if not frame.empty and 'date' in frame.columns:
        frame['date'] = pd.to_datetime(frame['date'], errors='coerce')

deepdive_output = CASE_DIR / 'deepdive'
deepdive_output.mkdir(parents=True, exist_ok=True)
```

## 2. (1) Regime bucket: direction regime


```python
# Use factor dates as evaluation dates if IC/L/S returns are empty.
seed_dates = labels_df[['date']].drop_duplicates().copy()
seed_dates['ic'] = float('nan')
seed_ic = ic_df[['date', 'ic']].copy() if {'date', 'ic'}.issubset(ic_df.columns) else seed_dates
if seed_ic.empty:
    seed_ic = seed_dates

seed_ls = (
    long_short_df[['date', 'long_short_return']].copy()
    if {'date', 'long_short_return'}.issubset(long_short_df.columns)
    else seed_dates.assign(long_short_return=float('nan'))
)

if prices.empty or not {'date', 'asset', 'close'}.issubset(prices.columns):
    regime_conditional = pd.DataFrame(columns=['date', 'bucket', 'ic', 'n'])
    print('Regime skipped: prices missing date/asset/close')
else:
    regime_df = classify_market_regimes(ic_df=seed_ic, long_short_df=seed_ls, prices=prices)
    if regime_df.empty:
        regime_conditional = pd.DataFrame(columns=['date', 'bucket', 'ic', 'n'])
        print('Regime frame empty after classification')
    else:
        regime_bucket = regime_df[['date', 'direction_regime']].rename(columns={'direction_regime': 'bucket'})
        regime_bucket = regime_bucket.dropna(subset=['bucket'])
        regime_conditional = conditional_ic_by_bucket(
            factor_df=factor_df,
            labels_df=labels_df,
            bucket_df=regime_bucket,
            group_col='bucket',
        )
        regime_conditional = regime_conditional.copy()
        regime_conditional['source'] = 'direction_regime'

regime_conditional.to_csv(deepdive_output / 'conditional_ic_by_regime.csv', index=False)
display(regime_conditional.head())
plot_bucket_series(regime_conditional[['date', 'bucket', 'ic']], title='Conditional IC by market direction regime')
```

## 3. (2) Size/liquidity/分位分层 (通用分层接口)


```python
size_cols = ['market_cap_log', 'market_cap', 'circ_mv', 'total_mv']
liquidity_cols = ['dollar_volume', 'amount', 'turnover', 'adv']
size_bucket_df = pd.DataFrame()
liquidity_bucket_df = pd.DataFrame()
size_source = ''
liquidity_source = ''

prices_df = prices if isinstance(prices, pd.DataFrame) else pd.DataFrame()
if prices_df.empty:
    print('Size/liquidity skipped: prices is not a DataFrame (or empty).')

for col in size_cols:
    if {'date', 'asset', col}.issubset(prices_df.columns):
        size_bucket_df = build_quantile_buckets(
            prices_df[['date', 'asset', col]].rename(columns={col: 'size_metric'}),
            value_col='size_metric',
            bucket_col='bucket',
            n_buckets=5,
        )
        size_source = col
        break

for col in liquidity_cols:
    if {'date', 'asset', col}.issubset(prices_df.columns):
        liquidity_bucket_df = build_quantile_buckets(
            prices_df[['date', 'asset', col]].rename(columns={col: 'liquidity_metric'}),
            value_col='liquidity_metric',
            bucket_col='bucket',
            n_buckets=3,
        )
        liquidity_source = col
        break

if size_bucket_df.empty:
    size_conditional = pd.DataFrame(columns=['date', 'bucket', 'ic', 'n'])
else:
    size_conditional = conditional_ic_by_bucket(
        factor_df=factor_df,
        labels_df=labels_df,
        bucket_df=size_bucket_df,
        group_col='bucket',
    )
    size_conditional['source'] = f'size:{size_source}'

if liquidity_bucket_df.empty:
    liquidity_conditional = pd.DataFrame(columns=['date', 'bucket', 'ic', 'n'])
else:
    liquidity_conditional = conditional_ic_by_bucket(
        factor_df=factor_df,
        labels_df=labels_df,
        bucket_df=liquidity_bucket_df,
        group_col='bucket',
    )
    liquidity_conditional['source'] = f'liquidity:{liquidity_source}'

size_conditional.to_csv(deepdive_output / 'conditional_ic_by_size.csv', index=False)
liquidity_conditional.to_csv(deepdive_output / 'conditional_ic_by_liquidity.csv', index=False)
plot_bucket_series(size_conditional, title='Conditional IC by size bucket')
plot_bucket_series(liquidity_conditional, title='Conditional IC by liquidity bucket')
```

## 4. (4) Asym-specific trailing-return buckets (past_ret_20 / recent_ret_3)


```python
trailing_candidates = ['past_ret_20', 'recent_ret_3']
trailing_frames = []

for col in trailing_candidates:
    if {'date', 'asset', col}.issubset(prices.columns):
        bucket_df = build_quantile_buckets(
            prices[['date', 'asset', col]].rename(columns={col: 'ret_metric'}),
            value_col='ret_metric',
            bucket_col='bucket',
            n_buckets=3,
        )
        result = conditional_ic_by_bucket(
            factor_df=factor_df,
            labels_df=labels_df,
            bucket_df=bucket_df,
            group_col='bucket',
        )
        result['source'] = col
        trailing_frames.append(result)

if trailing_frames:
    trailing_conditional = pd.concat(trailing_frames, ignore_index=True)
else:
    trailing_conditional = pd.DataFrame(columns=['date', 'bucket', 'ic', 'n', 'source'])

trailing_conditional.to_csv(deepdive_output / 'conditional_ic_by_trailing_return.csv', index=False)
if not trailing_conditional.empty:
    for source_name, block in trailing_conditional.groupby('source', sort=False):
        plot_bucket_series(block[['date', 'bucket', 'ic']], title=f'Conditional IC by trailing bucket: {source_name}')
else:
    print('No trailing return columns found (past_ret_20 or recent_ret_3).')
```
